```python
import json

def get_transcript_json(path):
    with open(path + "/transcript.json", "r", encoding="utf-8") as f:
        return json.load(f)

def save_annotated_transcript(path, data):
    with open(path + "/annotated_transcript.txt", "w", encoding="utf-8") as f:
        f.write(data)

def get_transcript(path):
    with open(path + "/transcript.txt", "r", encoding="utf-8") as f:
        return f.read()

def get_first_summaries(path):
    with open(path + "/first_summary.json", "r", encoding="utf-8") as f:
        return json.load(f)

def get_final_summary(path):
    with open(path + "/summary.md", "r", encoding="utf-8") as f:
        return f.read()

def save_qa(path, data):
    with open(path + "/data.jsonl", "w", encoding="utf-8") as f:
        for item in data:
            f.write(json.dumps(item, ensure_ascii=False) + "\n")
```

```python
import random


def data_to_qa(path):
    transcript = get_transcript(path)
    first_summaries = get_first_summaries(path)
    final_summary = get_final_summary(path)

    qa_pairs = []
    #qa_pairs += random.sample(first_summaries, k=1)
    
    final_prompt = "summarize point form: \n" + "\n".join([summary["completion"] for summary in first_summaries])
    qa_pairs += [{"prompt": final_prompt, "completion": final_summary}] * 500    
    save_qa(path, qa_pairs)
```

```python
from pathlib import Path

base = Path("data")

paths = [
    str(p)
    for p in base.glob("*/*/*")
    if p.is_dir()
]

print(paths)
```

```python
for path in paths:
    data_to_qa(path)
```

```python
def create_annotated_transcript(path):
    transcript_json = get_transcript_json(path)
    transcript = "\n".join([segment["speaker"] + ": " + segment["text"] for segment in transcript_json])
    save_annotated_transcript(path, transcript)
```
