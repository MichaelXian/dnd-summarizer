```python
import json
import string

stopwords = ["", "I", "a", "thank you", "yeah", "nice", "okay", "oh", "i love it", "good luck", "good", "well", "uh"]


def clean_transcript(path):
    with open(path + "/transcript.json", "r") as f:
        transcript = json.load(f)
        transcript = [item for item in transcript if item["text"].strip(string.punctuation).strip().lower() not in stopwords]
        
    with open(path + "/transcript.json", "w") as f:
        json.dump(transcript, f)
```

```python
from pathlib import Path

base = Path("data")

paths = [
    str(p)
    for p in base.glob("*/*/*")
    if p.is_dir()
]

print(paths)
```

```python
for path in paths:
    clean_transcript(path)
```
